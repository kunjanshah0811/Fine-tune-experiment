## Generate System Prompt with available tools

```python
import pandas as pd
import json
import random

# File paths
SYSTEM_PROMPT_PATH = "D:\\Uni Study\\Sem-6\\6.My Code\\2.Multi_turn\\system.md"
EXCEL_PATH = "D:\\Uni Study\\Sem-6\\6.My Code\\2.Multi_turn\\base_query_dataset.xlsx"
JSON_PATH = "D:\\Uni Study\\Sem-6\\6.My Code\\2.Multi_turn\\15_tools_no_category.json"
OUTPUT_EXCEL_PATH = "D:\\Uni Study\\Sem-6\\6.My Code\\2.Multi_turn\\base_system_prompts_output.xlsx"

# Read system prompt template
with open(SYSTEM_PROMPT_PATH, 'r') as f:
    system_prompt_template = f.read()

# Read JSON file with tools (flat array)
with open(JSON_PATH, 'r') as f:
    tools_data = json.load(f)

# Read Excel file
df = pd.read_excel(EXCEL_PATH)

# Function to find a tool by name in the flat array
def find_tool_by_name(tool_name):
    for tool in tools_data:
        if tool.get('function', {}).get('name') == tool_name.strip():
            return tool
    return None

# Function to get a random tool from all tools
def get_random_tool():
    return random.choice(tools_data)

# Function to extract tools based on tool_name or select random tools
def get_tools_for_query(tool_name):
    selected_tools = []
    
    if pd.isna(tool_name):  # If no tool_name, pick 3 random tools
        selected_tools = random.sample(tools_data, min(3, len(tools_data)))
    else:
        # Strip whitespace from tool_name and split by comma
        tool_names = [name.strip() for name in tool_name.split(',')]
        
        # Find and add the specified tools
        for name in tool_names:
            tool = find_tool_by_name(name)
            if tool:
                selected_tools.append(tool)
            else:
                print(f"Tool '{name}' not found")
        
        # Add random tools to make total of 3
        while len(selected_tools) < 3:
            random_tool = get_random_tool()
            # Make sure we don't add duplicate
            if random_tool not in selected_tools:
                selected_tools.append(random_tool)
    
    # Ensure we have exactly 3 tools
    if len(selected_tools) > 3:
        selected_tools = selected_tools[:3]
    
    # Format tools as JSON string
    return json.dumps(selected_tools, separators=(',', ': '))

# Generate system prompts with only the tools
system_prompts = []
tools_used = []  # Track which tools were included

for _, row in df.iterrows():
    tool_name = row['tool_name']
    
    # Get tools for the query
    tools_json = get_tools_for_query(tool_name)
    
    # Parse the JSON to extract tool names for tracking
    tools_list = json.loads(tools_json)
    tool_names = [tool.get('function', {}).get('name', 'unknown') for tool in tools_list]
    tools_used.append(', '.join(tool_names))
    
    # Replace {tools} placeholder with the actual tools JSON (no extra tags needed)
    system_prompt = system_prompt_template.replace("{tools}", tools_json)
    
    system_prompts.append(system_prompt)

# Create new DataFrame with id, Query, System_Prompt and tools
# Use the first available column that might contain queries
query_column = None
possible_query_columns = ['Query', 'query', 'question', 'Question', 'text', 'Text', 'prompt', 'Prompt']

for col in possible_query_columns:
    if col in df.columns:
        query_column = col
        break

if query_column is None:
    # If no recognized query column, use the second column (assuming first is id)
    query_column = df.columns[1] if len(df.columns) > 1 else df.columns[0]
    print(f"Using column '{query_column}' as query column")

output_df = pd.DataFrame({
    'id': df['id'],
    'Query': df[query_column],
    'System_Prompt': system_prompts,
    'Tools_Used': tools_used
})

# Save to new Excel file
output_df.to_excel(OUTPUT_EXCEL_PATH, index=False)

print(f"System prompts generated and saved to {OUTPUT_EXCEL_PATH}")
```

## Conversation - role (system,user,assistant & content)

```python
import pandas as pd
import json

# File paths
EXCEL_PATH = "D:\\Uni Study\\Sem-6\\6.My Code\\2.Multi_turn\\miss_para_query_dataset.xlsx"
OUTPUT_EXCEL_PATH = "D:\\Uni Study\\Sem-6\\6.My Code\\2.Multi_turn\\miss_para__conversation_output.xlsx"

# Read Excel file
df = pd.read_excel(EXCEL_PATH)

# Function to format multi-turn conversation
def create_conversation(row):
    system_prompt = row['System_Prompt']
    roles_data = row['Roles']
    
    # Parse the Roles column (assuming it's JSON string with user/assistant alternating)
    try:
        parsed_roles = json.loads(roles_data)
    except (json.JSONDecodeError, TypeError):
        # If parsing fails, return empty conversation with just system prompt
        return json.dumps([{"role": "system", "content": system_prompt}], indent=2, ensure_ascii=False)
    
    # Start conversation with system prompt
    conversation = [{"role": "system", "content": system_prompt}]
    
    # Add user and assistant messages from Roles column
    if isinstance(parsed_roles, list):
        conversation.extend(parsed_roles)
    elif isinstance(parsed_roles, dict):
        # If it's a single message, add it
        conversation.append(parsed_roles)
    
    # Convert to JSON string for Excel storage
    return json.dumps(conversation, indent=2, ensure_ascii=False)

# Generate conversations
df['Conversation'] = df.apply(create_conversation, axis=1)

# Save to new Excel file
df.to_excel(OUTPUT_EXCEL_PATH, index=False)

print(f"Multi-turn conversations generated and saved to {OUTPUT_EXCEL_PATH}")
```

```python
######## With Splitting
import pandas as pd
from datasets import Dataset
from huggingface_hub import login
import numpy as np
import sys

# File paths
EXCEL_PATH = "D:\\Uni Study\\Sem-6\\6.My Code\\2.Multi_turn\\2(1)Multi-turn.xlsx"

# Step 1: Read the Excel file
df = pd.read_excel(EXCEL_PATH)
total_rows = len(df)
print(f"Total rows in dataset: {total_rows}")

# Step 2: Shuffle and split the dataset
df_shuffled = df.sample(frac=1, random_state=42).reset_index(drop=True)

# Split into train, validation, and test sets (80% train, 10% validation, 10% test)
train_frac = 0.8
valid_frac = 0.1
test_frac = 0.1

# Calculate exact split sizes to ensure they sum to total rows
train_size = int(train_frac * total_rows)
valid_size = int(valid_frac * total_rows)
# Ensure test gets remaining rows so the total is exact
test_size = total_rows - train_size - valid_size

# Create train, validation, and test dataframes
df_train = df_shuffled[:train_size]
df_valid = df_shuffled[train_size:train_size + valid_size]
df_test = df_shuffled[train_size + valid_size:]

# Validation check
actual_total = len(df_train) + len(df_valid) + len(df_test)
print(f"Validation: Expected {total_rows} rows, got {actual_total} rows")

# Calculate memory usage
def get_size_in_mb(df):
    return df.memory_usage(deep=True).sum() / (1024 * 1024)

train_mb = get_size_in_mb(df_train)
valid_mb = get_size_in_mb(df_valid)
test_mb = get_size_in_mb(df_test)
total_mb = train_mb + valid_mb + test_mb

# Print detailed split information
print(f"Training set: {len(df_train)} rows ({len(df_train)/total_rows:.1%}), {train_mb:.2f} MB")
print(f"Validation set: {len(df_valid)} rows ({len(df_valid)/total_rows:.1%}), {valid_mb:.2f} MB")
print(f"Test set: {len(df_test)} rows ({len(df_test)/total_rows:.1%}), {test_mb:.2f} MB")
print(f"Total dataset size: {total_mb:.2f} MB")

# # Step 3: Convert each DataFrame to Hugging Face Dataset
# relevant_columns = ['id', 'Turn Type', 'Prompt_type', 'tool_needed', 'tool_call', 'tool_name', 
#                      'Query', 'Ground_truth', 'Missing_para', 'tools_in_system_prompt', 'System_Prompt', 'Conversation']

train_dataset = Dataset.from_pandas(df_train)
valid_dataset = Dataset.from_pandas(df_valid)
test_dataset = Dataset.from_pandas(df_test)

# Step 4: Create a DatasetDict for the split datasets
from datasets import DatasetDict

dataset_dict = DatasetDict({
    'train': train_dataset,
    'validation': valid_dataset,
    'test': test_dataset
})

# Print dataset information
print("\nDataset splits information:")
print(f"Train: {len(train_dataset)} examples")
print(f"Validation: {len(valid_dataset)} examples")
print(f"Test: {len(test_dataset)} examples")

# Step 5: Log in to Hugging Face Hub
login(token="hf_token")  # Use your actual token with write access

# Step 6: Create dataset card with metadata and statistics
dataset_description = f"""
# Single Turn Function Calling Dataset

## Dataset Description
This dataset contains {total_rows} examples of single-turn conversations for function calling with tool use.

## Dataset Statistics
- **Total examples**: {total_rows}
- **Train split**: {len(train_dataset)} examples ({len(df_train)/total_rows:.1%})
- **Validation split**: {len(valid_dataset)} examples ({len(df_valid)/total_rows:.1%})
- **Test split**: {len(test_dataset)} examples ({len(df_test)/total_rows:.1%})

## Dataset Size
- **Total size**: {total_mb:.2f} MB
- **Training set**: {train_mb:.2f} MB
- **Validation set**: {valid_mb:.2f} MB
- **Test set**: {test_mb:.2f} MB

## Dataset Features
- `id`: Unique identifier for each example
- `Turn Type`: Type of conversation turn
- `Prompt_type`: Type of prompt used
- `tool_needed`: Whether a tool is needed (boolean)
- `tool_call`: Tool call information
- `tool_name`: Name of the tool used
- `Query`: User query
- `Ground_truth`: Expected response
- `Missing_para`: Any missing parameters
- `tools_in_system_prompt`: Tools available in the system prompt
- `System_Prompt`: System prompt used
- `Conversation`: Full conversation including system, user, and assistant messages

## Tool Distribution
{df['tool_name'].value_counts().head(10).to_markdown()}

## License
This dataset is provided for research purposes.
"""

# Step 7: Push dataset to Hugging Face Hub with the split configuration and metadata
dataset_dict.push_to_hub(
    "kunjanshah/new_multi_turn_function_calling_split", 
    private=False
)
print("\nSplit dataset successfully pushed to Hugging Face Hub")
```

```python
from datasets import load_dataset
import pandas as pd
import os

def download_and_save_dataset(dataset_name, output_dir):
    """
    Download dataset from Hugging Face and save splits as Excel files
    """
    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)
    
    # Download dataset splits
    splits = ["train", "validation", "test"]
    
    for split in splits:
        print(f"Downloading {split} split...")
        
        # Load specific split
        dataset = load_dataset(dataset_name, split=split)
        
        # Convert to pandas DataFrame
        df = dataset.to_pandas()
        
        # Save to Excel
        output_path = os.path.join(output_dir, f"{split}_data.xlsx")
        df.to_excel(output_path, index=False)
        print(f"✅ Saved {split} split to: {output_path}")
        print(f"📊 Number of rows: {len(df)}")

if __name__ == "__main__":
    # Dataset parameters
    DATASET_NAME = "kunjanshah/new_multi_turn_function_calling_split"
    OUTPUT_DIR = "D:\\Uni Study\\Sem-6\\6.My Code\\2.Multi_turn\\dataset_splits"
    
    try:
        download_and_save_dataset(DATASET_NAME, OUTPUT_DIR)
        print("\n✨ All splits downloaded and saved successfully!")
        
    except Exception as e:
        print(f"❌ Error downloading dataset: {str(e)}")
```
