```python
import json
import pandas as pd
import random
from itertools import combinations
import re

def load_tools_from_file(filename):
    """Load tools from a JSON file"""
    try:
        with open(filename, 'r') as file:
            return json.load(file)
    except FileNotFoundError:
        print(f"Error: {filename} not found. Please create the file with your tools data.")
        return None

def generate_sample_data():
    """Generate sample data for various scenarios"""
    employees = ["EMP001", "EMP002", "EMP003", "EMP456", "EMP789"]
    customers = ["CUST001", "CUST002", "CUST003", "CUST456", "CUST789"]
    products = ["PROD001", "PROD002", "PROD003", "PROD456", "PROD789"]
    vendors = ["VEND001", "VEND002", "VEND003"]
    rooms = ["CONF-A-301", "CONF-B-202", "BOARD-001", "MEET-101"]
    
    return {
        "employees": employees,
        "customers": customers,
        "products": products,
        "vendors": vendors,
        "rooms": rooms
    }

def generate_query_and_params_for_tool(tool_name, tool_params, sample_data):
    """Generate realistic user queries with extractable parameters for specific tools"""
    
    # Generate specific data for each query
    employee_id = random.choice(sample_data['employees'])
    customer_id = random.choice(sample_data['customers'])
    product_id = random.choice(sample_data['products'])
    vendor_id = random.choice(sample_data['vendors'])
    room_id = random.choice(sample_data['rooms'])
    
    queries_with_params = {
        "get_employee_profile": {
            "query": f"I need to review the profile of employee {employee_id}. Can you retrieve their complete information including personal details, employment history, and performance metrics?",
            "params": {"employee_id": employee_id}
        },
        
        "create_sales_opportunity": {
            "query": f"I want to create a new sales opportunity for customer {customer_id} called 'Q4 Software License Deal' with an estimated value of $50,000 and 75% probability to close by 2024-12-31.",
            "params": {
                "customer_id": customer_id,
                "opportunity_name": "Q4 Software License Deal",
                "estimated_value": 50000,
                "probability": 75,
                "expected_close_date": "2024-12-31"
            }
        },
        
        "update_inventory_count": {
            "query": f"I need to update inventory for product {product_id}. We sold 15 units today, so please reduce the count by 15 and mark it as a sale. Updated by {employee_id}.",
            "params": {
                "product_id": product_id,
                "quantity_change": -15,
                "reason": "sale",
                "updated_by": employee_id
            }
        },
        
        "process_invoice_payment": {
            "query": f"I need to process payment for invoice INV-2024-001 from vendor {vendor_id} for $15,750 in USD using ACH transfer.",
            "params": {
                "invoice_id": "INV-2024-001",
                "vendor_id": vendor_id,
                "amount": 15750,
                "currency": "USD",
                "payment_method": "ACH"
            }
        },
        
        "create_helpdesk_ticket": {
            "query": f"Employee {employee_id} is having laptop issues - the screen keeps flickering. Create a hardware ticket with high priority. The laptop is a Dell XPS 15 purchased last year.",
            "params": {
                "user_id": employee_id,
                "issue_summary": "Laptop screen flickering",
                "category": "hardware",
                "priority": "high",
                "description": "The laptop screen keeps flickering. The laptop is a Dell XPS 15 purchased last year."
            }
        },
        
        "approve_procurement_request": {
            "query": f"I need to approve procurement request REQ-2024-056 as the approving manager {employee_id}. The request looks good and is within budget.",
            "params": {
                "request_id": "REQ-2024-056",
                "approver_id": employee_id,
                "status": "Approved",
                "budget_check": True
            }
        },
        
        "create_sales_proposal": {
            "query": f"Generate a sales proposal for client {customer_id} titled 'Enterprise Software Solution'. Include product {product_id} with quantity 5 at $2,000 per unit, 30-day payment terms.",
            "params": {
                "client_id": customer_id,
                "proposal_title": "Enterprise Software Solution",
                "products": [{
                    "product_id": product_id,
                    "quantity": 5,
                    "unit_price": 2000
                }],
                "payment_terms": {"due_in_days": 30}
            }
        },
        
        "conduct_compliance_audit": {
            "query": f"Initiate compliance audit AUDIT-2024-Q3 for GDPR and HIPAA frameworks covering IT and Finance departments from 2024-07-01 to 2024-09-30. Lead auditor is {employee_id} focusing on data privacy and security risks with $50,000 materiality threshold.",
            "params": {
                "audit_id": "AUDIT-2024-Q3",
                "regulatory_frameworks": ["GDPR", "HIPAA"],
                "business_units": ["IT", "Finance"],
                "audit_period": {
                    "start_date": "2024-07-01",
                    "end_date": "2024-09-30"
                },
                "lead_auditor": employee_id,
                "risk_categories": ["data_privacy", "security"],
                "materiality_threshold": 50000
            }
        },
        
        "book_meeting_room": {
            "query": f"Book conference room {room_id} for 2024-10-15 from 14:00 to 16:00 for 8 attendees. Meeting title 'Quarterly Review' organized by {employee_id}. We'll need a projector.",
            "params": {
                "room_id": room_id,
                "date": "2024-10-15",
                "start_time": "14:00",
                "end_time": "16:00",
                "attendee_count": 8,
                "meeting_title": "Quarterly Review",
                "organizer_id": employee_id,
                "equipment_needed": ["projector"]
            }
        },
        
        "track_shipment": {
            "query": "Track shipment 1Z999AA1234567890 with UPS carrier and include the full tracking history.",
            "params": {
                "tracking_number": "1Z999AA1234567890",
                "carrier": "UPS",
                "include_history": True
            }
        },
        
        "escalate_support_ticket": {
            "query": "Escalate support ticket TKT-2024-001 to engineering team due to critical system outage. Please notify the customer about the escalation.",
            "params": {
                "ticket_id": "TKT-2024-001",
                "reason": "critical",
                "target_team": "engineering",
                "notify_customer": True
            }
        },
        
        "create_performance_review": {
            "query": f"Create annual performance review for employee {employee_id} with reviewer {customer_id} covering period 2024-01-01 to 2024-12-31. Overall rating is 4 out of 5.",
            "params": {
                "employee_id": employee_id,
                "reviewer_id": customer_id,
                "review_period": {
                    "start_date": "2024-01-01",
                    "end_date": "2024-12-31"
                },
                "overall_rating": 4,
                "review_type": "annual"
            }
        },
        
        "schedule_meeting": {
            "query": "Schedule a meeting titled 'Project Kickoff' from 2024-10-15T14:00:00 to 2024-10-15T16:00:00 with required participants john@company.com and sarah@company.com, optional participant mike@company.com. Agenda: Discuss project timeline and deliverables.",
            "params": {
                "title": "Project Kickoff",
                "start_time": "2024-10-15T14:00:00",
                "end_time": "2024-10-15T16:00:00",
                "participants": [
                    {"email": "john@company.com", "role": "required"},
                    {"email": "sarah@company.com", "role": "required"},
                    {"email": "mike@company.com", "role": "optional"}
                ],
                "agenda": "Discuss project timeline and deliverables"
            }
        },
        
        "track_ad_performance": {
            "query": "Track performance for Google Ads campaigns AD-001 and AD-002, monitoring impressions, clicks, and cost per click metrics. Refresh data every 30 minutes.",
            "params": {
                "ad_ids": ["AD-001", "AD-002"],
                "platform": "google_ads",
                "metrics": ["impressions", "clicks", "cost_per_click"],
                "polling_interval_minutes": 30
            }
        },
        
        "send_offer_letter": {
            "query": "Send job offer to candidate CAND-2024-001 for Senior Developer position with $95,000 base salary, start date 2024-11-01. Offer expires in 7 days.",
            "params": {
                "candidate_id": "CAND-2024-001",
                "job_title": "Senior Developer",
                "salary_base": 95000,
                "start_date": "2024-11-01",
                "expiration_days": 7
            }
        }
    }
    
    # Return default if tool not found
    default_query = f"Please help me with {tool_name} functionality."
    return queries_with_params.get(tool_name, {"query": default_query, "params": {}})

def generate_ground_truth(tool_name, extracted_params):
    """Generate ground truth tool calls using extracted parameters"""
    return {
        "name": tool_name,
        "arguments": extracted_params
    }

def generate_query_dataset(tools_data, num_rows=50):
    """Generate dataset with tool pairs and corresponding queries"""
    
    if not tools_data:
        return []
    
    sample_data = generate_sample_data()
    dataset = []
    
    # Extract tool names and definitions
    tools_dict = {}
    for tool in tools_data:
        tool_name = tool['function']['name']
        tools_dict[tool_name] = tool['function']
    
    tool_names = list(tools_dict.keys())
    
    # Generate combinations of 2 tools
    tool_pairs = list(combinations(tool_names, 2))
    
    for i in range(num_rows):
        # Select a random tool pair
        tool_pair = random.choice(tool_pairs)
        tool1, tool2 = tool_pair
        
        # Generate queries and parameters for each tool
        query_data1 = generate_query_and_params_for_tool(tool1, tools_dict[tool1], sample_data)
        query_data2 = generate_query_and_params_for_tool(tool2, tools_dict[tool2], sample_data)
        
        # Extract query text and parameters
        query1 = query_data1["query"]
        query2 = query_data2["query"]
        params1 = query_data1["params"]
        params2 = query_data2["params"]
        
        # Generate ground truth using extracted parameters
        ground_truth1 = generate_ground_truth(tool1, params1)
        ground_truth2 = generate_ground_truth(tool2, params2)
        
        # Create formatted queries as JSON objects
        formatted_queries = [
            {"role": "user", "content": query1},
            {"role": "user", "content": query2}
        ]
        
        # Create dataset entry
        row = {
            'id': f"query_pair_{i+1:03d}",
            'tools_used': f"{tool1}, {tool2}",
            'queries': json.dumps(formatted_queries, indent=2),
            'ground_truth_1': json.dumps(ground_truth1, indent=2),
            'ground_truth_2': json.dumps(ground_truth2, indent=2)
        }
        
        dataset.append(row)
    
    return dataset

def save_to_excel(dataset, filename='query_dataset.xlsx'):
    """Save dataset to Excel file"""
    
    # Prepare data for Excel
    excel_data = []
    for row in dataset:
        excel_row = {
            'ID': row['id'],
            'Tools Used': row['tools_used'],
            'Queries (2 queries in JSON format)': row['queries'],
            'Ground Truth 1': row['ground_truth_1'],
            'Ground Truth 2': row['ground_truth_2']
        }
        excel_data.append(excel_row)
    
    # Create DataFrame and save to Excel
    df = pd.DataFrame(excel_data)
    
    # Adjust column widths for better readability
    with pd.ExcelWriter(filename, engine='openpyxl') as writer:
        df.to_excel(writer, index=False, sheet_name='Query Dataset')
        
        # Get the workbook and worksheet
        workbook = writer.book
        worksheet = writer.sheets['Query Dataset']
        
        # Set column widths
        worksheet.column_dimensions['A'].width = 15  # ID
        worksheet.column_dimensions['B'].width = 30  # Tools Used
        worksheet.column_dimensions['C'].width = 80  # Queries
        worksheet.column_dimensions['D'].width = 50  # Ground Truth 1
        worksheet.column_dimensions['E'].width = 50  # Ground Truth 2
        
        # Enable text wrapping for better readability
        from openpyxl.styles import Alignment
        for row in worksheet.iter_rows():
            for cell in row:
                cell.alignment = Alignment(wrap_text=True, vertical='top')
    
    print(f"Dataset saved to {filename}")
    print(f"Generated {len(excel_data)} query pairs")

def main():
    """Main function to generate and save query dataset"""
    
    # Load tools from JSON file
    tools_filename = 'tools.json'  # Make sure this file exists with your tools data
    tools_data = load_tools_from_file(tools_filename)
    
    if tools_data is None:
        # Create sample tools.json file if it doesn't exist
        print("Creating sample tools.json file...")
        sample_tools = [
            {
                "type": "function",
                "function": {
                    "name": "get_employee_profile", 
                    "description": "Retrieve employee profile information",
                    "parameters": {
                        "type": "object",
                        "properties": {
                            "employee_id": {"type": "string", "description": "Employee ID"}
                        },
                        "required": ["employee_id"]
                    }
                }
            }
        ]
        
        with open(tools_filename, 'w') as f:
            json.dump(sample_tools, f, indent=2)
        
        print(f"Please update {tools_filename} with your complete tools data and run again.")
        return
    
    # Generate dataset
    print("Generating query dataset...")
    dataset = generate_query_dataset(tools_data, num_rows=50)
    
    # Save to Excel
    save_to_excel(dataset, 'query1_dataset.xlsx')
    
    # Display sample
    print("\nSample generated data:")
    for i, row in enumerate(dataset[:2]):
        print(f"\nRow {i+1}:")
        print(f"ID: {row['id']}")
        print(f"Tools: {row['tools_used']}")
        
        # Parse and display queries nicely
        queries = json.loads(row['queries'])
        print(f"Query 1: {queries[0]['content'][:100]}...")
        print(f"Query 2: {queries[1]['content'][:100]}...")
        
        # Show ground truth structure
        gt1 = json.loads(row['ground_truth_1'])
        print(f"Ground Truth 1 Tool: {gt1['name']}")

if __name__ == "__main__":
    main()
```

```python
import json
import pandas as pd
import random
from itertools import combinations
import re

def load_tools_from_file(filename):
    """Load tools from a JSON file"""
    try:
        with open(filename, 'r') as file:
            return json.load(file)
    except FileNotFoundError:
        print(f"Error: {filename} not found. Please create the file with your tools data.")
        return None

def generate_sample_data():
    """Generate sample data for various scenarios"""
    employees = ["EMP001", "EMP002", "EMP003", "EMP456", "EMP789"]
    customers = ["CUST001", "CUST002", "CUST003", "CUST456", "CUST789"]
    products = ["PROD001", "PROD002", "PROD003", "PROD456", "PROD789"]
    vendors = ["VEND001", "VEND002", "VEND003"]
    rooms = ["CONF-A-301", "CONF-B-202", "BOARD-001", "MEET-101"]
    
    return {
        "employees": employees,
        "customers": customers,
        "products": products,
        "vendors": vendors,
        "rooms": rooms
    }

def generate_query_and_params_for_tool(tool_name, tool_params, sample_data, missing_param=False):
    """Generate realistic user queries with extractable parameters for specific tools, optionally missing a required parameter"""
    
    # Generate specific data for each query
    employee_id = random.choice(sample_data['employees'])
    customer_id = random.choice(sample_data['customers'])
    product_id = random.choice(sample_data['products'])
    vendor_id = random.choice(sample_data['vendors'])
    room_id = random.choice(sample_data['rooms'])
    
    queries_with_params = {
        "get_employee_profile": {
            "query": "I need to review an employee's profile. Can you retrieve their complete information including personal details, employment history, and performance metrics?" if missing_param else f"I need to review the profile of employee {employee_id}. Can you retrieve their complete information including personal details, employment history, and performance metrics?",
            "params": {"employee_id": employee_id},
            "missing_param": "employee_id" if missing_param else None
        },
        
        "create_sales_opportunity": {
            "query": "I want to create a new sales opportunity called 'Q4 Software License Deal' with an estimated value of $50,000 and 75% probability to close by 2024-12-31." if missing_param else f"I want to create a new sales opportunity for customer {customer_id} called 'Q4 Software License Deal' with an estimated value of $50,000 and 75% probability to close by 2024-12-31.",
            "params": {
                "customer_id": customer_id,
                "opportunity_name": "Q4 Software License Deal",
                "estimated_value": 50000,
                "probability": 75,
                "expected_close_date": "2024-12-31"
            },
            "missing_param": "customer_id" if missing_param else None
        },
        
        "update_inventory_count": {
            "query": f"I need to update inventory. We sold 15 units today, so please reduce the count by 15 and mark it as a sale. Updated by {employee_id}." if missing_param else f"I need to update inventory for product {product_id}. We sold 15 units today, so please reduce the count by 15 and mark it as a sale. Updated by {employee_id}.",
            "params": {
                "product_id": product_id,
                "quantity_change": -15,
                "reason": "sale",
                "updated_by": employee_id
            },
            "missing_param": "product_id" if missing_param else None
        },
        
        "process_invoice_payment": {
            "query": "I need to process payment for invoice INV-2024-001 for $15,750 in USD using ACH transfer." if missing_param else f"I need to process payment for invoice INV-2024-001 from vendor {vendor_id} for $15,750 in USD using ACH transfer.",
            "params": {
                "invoice_id": "INV-2024-001",
                "vendor_id": vendor_id,
                "amount": 15750,
                "currency": "USD",
                "payment_method": "ACH"
            },
            "missing_param": "vendor_id" if missing_param else None
        },
        
        "create_helpdesk_ticket": {
            "query": "An employee is having laptop issues - the screen keeps flickering. Create a hardware ticket with high priority. The laptop is a Dell XPS 15 purchased last year." if missing_param else f"Employee {employee_id} is having laptop issues - the screen keeps flickering. Create a hardware ticket with high priority. The laptop is a Dell XPS 15 purchased last year.",
            "params": {
                "user_id": employee_id,
                "issue_summary": "Laptop screen flickering",
                "category": "hardware",
                "priority": "high",
                "description": "The laptop screen keeps flickering. The laptop is a Dell XPS 15 purchased last year."
            },
            "missing_param": "user_id" if missing_param else None
        },
        
        "approve_procurement_request": {
            "query": "I need to approve a procurement request as the approving manager. The request looks good and is within budget." if missing_param else f"I need to approve procurement request REQ-2024-056 as the approving manager {employee_id}. The request looks good and is within budget.",
            "params": {
                "request_id": "REQ-2024-056",
                "approver_id": employee_id,
                "status": "Approved",
                "budget_check": True
            },
            "missing_param": "approver_id" if missing_param else None
        },
        
        "create_sales_proposal": {
            "query": "Generate a sales proposal titled 'Enterprise Software Solution'. Include a product with quantity 5 at $2,000 per unit, 30-day payment terms." if missing_param else f"Generate a sales proposal for client {customer_id} titled 'Enterprise Software Solution'. Include product {product_id} with quantity 5 at $2,000 per unit, 30-day payment terms.",
            "params": {
                "client_id": customer_id,
                "proposal_title": "Enterprise Software Solution",
                "products": [{
                    "product_id": product_id,
                    "quantity": 5,
                    "unit_price": 2000
                }],
                "payment_terms": {"due_in_days": 30}
            },
            "missing_param": "client_id" if missing_param else None
        },
        
        "conduct_compliance_audit": {
            "query": "Initiate compliance audit AUDIT-2024-Q3 for GDPR and HIPAA frameworks covering IT and Finance departments from 2024-07-01 to 2024-09-30, focusing on data privacy and security risks with $50,000 materiality threshold." if missing_param else f"Initiate compliance audit AUDIT-2024-Q3 for GDPR and HIPAA frameworks covering IT and Finance departments from 2024-07-01 to 2024-09-30. Lead auditor is {employee_id} focusing on data privacy and security risks with $50,000 materiality threshold.",
            "params": {
                "audit_id": "AUDIT-2024-Q3",
                "regulatory_frameworks": ["GDPR", "HIPAA"],
                "business_units": ["IT", "Finance"],
                "audit_period": {
                    "start_date": "2024-07-01",
                    "end_date": "2024-09-30"
                },
                "lead_auditor": employee_id,
                "risk_categories": ["data_privacy", "security"],
                "materiality_threshold": 50000
            },
            "missing_param": "lead_auditor" if missing_param else None
        },
        
        "book_meeting_room": {
            "query": "Book a conference room for 2024-10-15 from 14:00 to 16:00 for 8 attendees. Meeting title 'Quarterly Review'. We'll need a projector." if missing_param else f"Book conference room {room_id} for 2024-10-15 from 14:00 to 16:00 for 8 attendees. Meeting title 'Quarterly Review' organized by {employee_id}. We'll need a projector.",
            "params": {
                "room_id": room_id,
                "date": "2024-10-15",
                "start_time": "14:00",
                "end_time": "16:00",
                "attendee_count": 8,
                "meeting_title": "Quarterly Review",
                "organizer_id": employee_id,
                "equipment_needed": ["projector"]
            },
            "missing_param": "room_id" if missing_param else None
        },
        
        "track_shipment": {
            "query": "Track a shipment with UPS carrier and include the full tracking history." if missing_param else "Track shipment 1Z999AA1234567890 with UPS carrier and include the full tracking history.",
            "params": {
                "tracking_number": "1Z999AA1234567890",
                "carrier": "UPS",
                "include_history": True
            },
            "missing_param": "tracking_number" if missing_param else None
        },
        
        "escalate_support_ticket": {
            "query": "Escalate a support ticket to engineering team due to critical system outage. Please notify the customer about the escalation." if missing_param else "Escalate support ticket TKT-2024-001 to engineering team due to critical system outage. Please notify the customer about the escalation.",
            "params": {
                "ticket_id": "TKT-2024-001",
                "reason": "critical",
                "target_team": "engineering",
                "notify_customer": True
            },
            "missing_param": "ticket_id" if missing_param else None
        },
        
        "create_performance_review": {
            "query": f"Create annual performance review with reviewer {customer_id} covering period 2024-01-01 to 2024-12-31. Overall rating is 4 out of 5." if missing_param else f"Create annual performance review for employee {employee_id} with reviewer {customer_id} covering period 2024-01-01 to 2024-12-31. Overall rating is 4 out of 5.",
            "params": {
                "employee_id": employee_id,
                "reviewer_id": customer_id,
                "review_period": {
                    "start_date": "2024-01-01",
                    "end_date": "2024-12-31"
                },
                "overall_rating": 4,
                "review_type": "annual"
            },
            "missing_param": "employee_id" if missing_param else None
        },
        
        "schedule_meeting": {
            "query": "Schedule a meeting titled 'Project Kickoff' from 2024-10-15T14:00:00 to 2024-10-15T16:00:00. Agenda: Discuss project timeline and deliverables." if missing_param else "Schedule a meeting titled 'Project Kickoff' from 2024-10-15T14:00:00 to 2024-10-15T16:00:00 with required participants john@company.com and sarah@company.com, optional participant mike@company.com. Agenda: Discuss project timeline and deliverables.",
            "params": {
                "title": "Project Kickoff",
                "start_time": "2024-10-15T14:00:00",
                "end_time": "2024-10-15T16:00:00",
                "participants": [
                    {"email": "john@company.com", "role": "required"},
                    {"email": "sarah@company.com", "role": "required"},
                    {"email": "mike@company.com", "role": "optional"}
                ],
                "agenda": "Discuss project timeline and deliverables"
            },
            "missing_param": "participants" if missing_param else None
        },
        
        "track_ad_performance": {
            "query": "Track performance for Google Ads campaigns, monitoring impressions, clicks, and cost per click metrics. Refresh data every 30 minutes." if missing_param else "Track performance for Google Ads campaigns AD-001 and AD-002, monitoring impressions, clicks, and cost per click metrics. Refresh data every 30 minutes.",
            "params": {
                "ad_ids": ["AD-001", "AD-002"],
                "platform": "google_ads",
                "metrics": ["impressions", "clicks", "cost_per_click"],
                "polling_interval_minutes": 30
            },
            "missing_param": "ad_ids" if missing_param else None
        },
        
        "send_offer_letter": {
            "query": "Send job offer for Senior Developer position with $95,000 base salary, start date 2024-11-01. Offer expires in 7 days." if missing_param else "Send job offer to candidate CAND-2024-001 for Senior Developer position with $95,000 base salary, start date 2024-11-01. Offer expires in 7 days.",
            "params": {
                "candidate_id": "CAND-2024-001",
                "job_title": "Senior Developer",
                "salary_base": 95000,
                "start_date": "2024-11-01",
                "expiration_days": 7
            },
            "missing_param": "candidate_id" if missing_param else None
        }
    }
    
    # Return default if tool not found
    default_query = f"Please help me with {tool_name} functionality."
    return queries_with_params.get(tool_name, {"query": default_query, "params": {}, "missing_param": None})

def generate_ground_truth(tool_name, extracted_params):
    """Generate ground truth tool calls using extracted parameters"""
    return {
        "name": tool_name,
        "arguments": extracted_params
    }

def generate_query_dataset(tools_data, num_rows=50):
    """Generate dataset with multi-turn queries for one tool, handling missing parameters"""
    
    if not tools_data:
        return []
    
    sample_data = generate_sample_data()
    dataset = []
    
    # Extract tool names and definitions
    tools_dict = {}
    for tool in tools_data:
        tool_name = tool['function']['name']
        tools_dict[tool_name] = tool['function']
    
    tool_names = list(tools_dict.keys())
    
    for i in range(num_rows):
        # Select a random tool
        tool = random.choice(tool_names)
        
        # Generate query with missing parameter
        query_data_missing = generate_query_and_params_for_tool(tool, tools_dict[tool], sample_data, missing_param=True)
        
        # Generate complete query parameters for ground truth
        query_data_complete = generate_query_and_params_for_tool(tool, tools_dict[tool], sample_data, missing_param=False)
        
        # Extract query text and parameters
        query_missing = query_data_missing["query"]
        missing_param = query_data_missing["missing_param"]
        params_complete = query_data_complete["params"]
        
        # Generate model's clarification request
        clarification_query = f"Please provide the missing {missing_param} to proceed with the {tool} request."
        
        # Generate user's response with the missing parameter
        user_response = f"The {missing_param} is {params_complete.get(missing_param, 'unknown')}."
        
        # Generate ground truth using complete parameters
        ground_truth = generate_ground_truth(tool, params_complete)
        
        # Create formatted multi-turn queries as JSON objects
        formatted_queries = [
            {"role": "user", "content": query_missing},
            {"role": "model", "content": clarification_query},
            {"role": "user", "content": user_response}
        ]
        
        # Create dataset entry
        row = {
            'id': f"query_{i+1:03d}",
            'tool_used': tool,
            'queries': json.dumps(formatted_queries, indent=2),
            'ground_truth': json.dumps(ground_truth, indent=2),
            'missing_param': missing_param or "None"
        }
        
        dataset.append(row)
    
    return dataset

def save_to_excel(dataset, filename='query_dataset.xlsx'):
    """Save dataset to Excel file"""
    
    # Prepare data for Excel
    excel_data = []
    for row in dataset:
        excel_row = {
            'ID': row['id'],
            'Tool Used': row['tool_used'],
            'Queries (3 turns in JSON format)': row['queries'],
            'Ground Truth': row['ground_truth'],
            'Missing Parameter': row['missing_param']
        }
        excel_data.append(excel_row)
    
    # Create DataFrame and save to Excel
    df = pd.DataFrame(excel_data)
    
    # Adjust column widths for better readability
    with pd.ExcelWriter(filename, engine='openpyxl') as writer:
        df.to_excel(writer, index=False, sheet_name='Query Dataset')
        
        # Get the workbook and worksheet
        workbook = writer.book
        worksheet = writer.sheets['Query Dataset']
        
        # Set column widths
        worksheet.column_dimensions['A'].width = 15  # ID
        worksheet.column_dimensions['B'].width = 30  # Tool Used
        worksheet.column_dimensions['C'].width = 80  # Queries
        worksheet.column_dimensions['D'].width = 50  # Ground Truth
        worksheet.column_dimensions['E'].width = 30  # Missing Parameter
        
        # Enable text wrapping for better readability
        from openpyxl.styles import Alignment
        for row in worksheet.iter_rows():
            for cell in row:
                cell.alignment = Alignment(wrap_text=True, vertical='top')
    
    print(f"Dataset saved to {filename}")
    print(f"Generated {len(excel_data)} query sets")

def main():
    """Main function to generate and save query dataset"""
    
    # Load tools from JSON file
    tools_filename = 'tools.json'  # Make sure this file exists with your tools data
    tools_data = load_tools_from_file(tools_filename)
    
    if tools_data is None:
        # Create sample tools.json file if it doesn't exist
        print("Creating sample tools.json file...")
        sample_tools = [
            {
                "type": "function",
                "function": {
                    "name": "get_employee_profile", 
                    "description": "Retrieve employee profile information",
                    "parameters": {
                        "type": "object",
                        "properties": {
                            "employee_id": {"type": "string", "description": "Employee ID"}
                        },
                        "required": ["employee_id"]
                    }
                }
            }
        ]
        
        with open(tools_filename, 'w') as f:
            json.dump(sample_tools, f, indent=2)
        
        print(f"Please update {tools_filename} with your complete tools data and run again.")
        return
    
    # Generate dataset
    print("Generating query dataset...")
    dataset = generate_query_dataset(tools_data, num_rows=50)
    
    # Save to Excel
    save_to_excel(dataset, 'query2_dataset.xlsx')
    
    # Display sample
    print("\nSample generated data:")
    for i, row in enumerate(dataset[:2]):
        print(f"\nRow {i+1}:")
        print(f"ID: {row['id']}")
        print(f"Tool: {row['tool_used']}")
        
        # Parse and display queries nicely
        queries = json.loads(row['queries'])
        print(f"Query 1 (User): {queries[0]['content'][:100]}...")
        print(f"Query 2 (Model): {queries[1]['content'][:100]}...")
        print(f"Query 3 (User): {queries[2]['content'][:100]}...")
        
        # Show ground truth structure
        gt = json.loads(row['ground_truth'])
        print(f"Ground Truth Tool: {gt['name']}")
        print(f"Missing Parameter: {row['missing_param']}")

if __name__ == "__main__":
    main()
```
