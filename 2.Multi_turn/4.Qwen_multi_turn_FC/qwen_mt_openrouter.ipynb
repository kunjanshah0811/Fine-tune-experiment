```python
# Enterprise Function Definitions with Mock Data

def get_employee_profile(employee_id: str):
    """Returns employee profile information including personal details and performance metrics.
    
    Required parameters:
    - employee_id: str - Unique identifier for the employee
    """
    employee_data = {
        "EMP001": {
            "name": "John Smith",
            "department": "Engineering",
            "position": "Senior Developer",
            "hire_date": "2020-03-15",
            "performance_rating": 4.2,
            "status": "Active"
        },
        "EMP002": {
            "name": "Sarah Johnson",
            "department": "Marketing",
            "position": "Marketing Manager",
            "hire_date": "2019-08-22",
            "performance_rating": 4.5,
            "status": "Active"
        }
    }
    return employee_data.get(employee_id, {
        "error": "Employee not found",
        "status": "Not Found"
    })

def create_sales_opportunity(customer_id: str, opportunity_name: str, estimated_value: float, probability: int, expected_close_date: str = None):
    """Create a new sales opportunity with lead information and initial assessment.
    
    Required parameters:
    - customer_id: str - Associated customer ID
    - opportunity_name: str - Name/title of the sales opportunity
    - estimated_value: float - Estimated deal value in USD
    - probability: int - Probability of closing the deal (0-100%)
    
    Optional parameters:
    - expected_close_date: str - Expected date to close the deal (YYYY-MM-DD)
    """
    opportunity_id = f"OPP_{customer_id}_{len(opportunity_name)}"
    
    return {
        "success": True,
        "opportunity_id": opportunity_id,
        "message": f"Sales opportunity '{opportunity_name}' created successfully",
        "customer_id": customer_id,
        "estimated_value": estimated_value,
        "probability": probability,
        "status": "Created"
    }

def update_inventory_count(product_id: str, quantity_change: int, reason: str, updated_by: str):
    """Update product inventory quantity with tracking information.
    
    Required parameters:
    - product_id: str - Product identifier
    - quantity_change: int - Quantity delta (positive/negative)
    - reason: str - Change reason (sale, purchase, return, damage, adjustment)
    - updated_by: str - Employee ID performing update
    """
    current_inventory = {
        "PROD001": 150,
        "PROD002": 75,
        "PROD003": 200
    }
    
    current_qty = current_inventory.get(product_id, 100)
    new_qty = current_qty + quantity_change
    
    return {
        "success": True,
        "product_id": product_id,
        "previous_quantity": current_qty,
        "quantity_change": quantity_change,
        "new_quantity": new_qty,
        "reason": reason,
        "updated_by": updated_by,
        "timestamp": "2025-09-13T10:30:00Z",
        "message": f"Inventory updated successfully for {product_id}"
    }

def process_invoice_payment(invoice_id: str, vendor_id: str, amount: float, currency: str = "USD", payment_method: str = "ACH"):
    """Process an invoice payment with vendor and amount details.
    
    Required parameters:
    - invoice_id: str - Invoice ID
    - vendor_id: str - Vendor ID
    - amount: float - Payment amount
    
    Optional parameters:
    - currency: str - Currency code (USD, EUR, GBP) - default: USD
    - payment_method: str - Payment method (ACH, Wire, Check) - default: ACH
    """
    transaction_id = f"TXN_{invoice_id}_{vendor_id}"
    
    return {
        "success": True,
        "transaction_id": transaction_id,
        "invoice_id": invoice_id,
        "vendor_id": vendor_id,
        "amount": amount,
        "currency": currency,
        "payment_method": payment_method,
        "status": "Processed",
        "processing_date": "2025-09-13",
        "message": f"Payment of {amount} {currency} processed successfully"
    }

def create_helpdesk_ticket(user_id: str, issue_summary: str, category: str, description: str, priority: str = "medium", attachments: list = None):
    """Create an IT support ticket for user issues.
    
    Required parameters:
    - user_id: str - Employee ID
    - issue_summary: str - Brief issue summary
    - category: str - Issue category (hardware, software, network, security, email, phone, printer, access)
    - description: str - Detailed problem description
    
    Optional parameters:
    - priority: str - Priority level (low, medium, high, critical) - default: medium
    - attachments: list - Attached files information
    """
    ticket_id = f"TICKET_{user_id}_{category.upper()}_{len(issue_summary)}"
    
    return {
        "success": True,
        "ticket_id": ticket_id,
        "user_id": user_id,
        "issue_summary": issue_summary,
        "category": category,
        "priority": priority,
        "status": "Open",
        "created_date": "2025-09-13T10:30:00Z",
        "estimated_resolution": "2025-09-15T10:30:00Z",
        "message": f"Support ticket {ticket_id} created successfully"
    }

def approve_procurement_request(request_id: str, approver_id: str, status: str, comments: str = None, budget_check: bool = True):
    """Approve or reject a procurement request with optional comments.
    
    Required parameters:
    - request_id: str - ID of the procurement request
    - approver_id: str - ID of the approving manager
    - status: str - Approval status (Approved, Rejected, Pending)
    
    Optional parameters:
    - comments: str - Optional comments for approval/rejection
    - budget_check: bool - Whether to enforce budget validation - default: True
    """
    return {
        "success": True,
        "request_id": request_id,
        "approver_id": approver_id,
        "status": status,
        "comments": comments or "No additional comments",
        "budget_check_passed": budget_check,
        "approval_date": "2025-09-13T10:30:00Z",
        "message": f"Procurement request {request_id} has been {status.lower()}"
    }

def create_sales_proposal(client_id: str, proposal_title: str, products: list, payment_terms: dict = None):
    """Generate a sales proposal for a client with products and terms.
    
    Required parameters:
    - client_id: str - Client ID
    - proposal_title: str - Proposal title
    - products: list - List of products with product_id, quantity, unit_price
    
    Optional parameters:
    - payment_terms: dict - Payment terms with due_in_days
    """
    proposal_id = f"PROP_{client_id}_{len(proposal_title)}"
    total_value = sum(p['quantity'] * p['unit_price'] for p in products)
    
    return {
        "success": True,
        "proposal_id": proposal_id,
        "client_id": client_id,
        "proposal_title": proposal_title,
        "total_value": total_value,
        "product_count": len(products),
        "payment_terms": payment_terms or {"due_in_days": 30},
        "created_date": "2025-09-13",
        "status": "Draft",
        "message": f"Sales proposal {proposal_id} created with total value ${total_value}"
    }

def conduct_compliance_audit(audit_id: str, regulatory_frameworks: list, business_units: list, audit_period: dict, lead_auditor: str, risk_categories: list, materiality_threshold: float = None, automated_testing: bool = True):
    """Execute a compliance audit with risk assessment and reporting.
    
    Required parameters:
    - audit_id: str - Unique identifier for the audit
    - regulatory_frameworks: list - Frameworks to audit against (GDPR, HIPAA, SOX, etc.)
    - business_units: list - Business units to include
    - audit_period: dict - Start and end dates for audit period
    - lead_auditor: str - Employee ID of lead auditor
    - risk_categories: list - Risk categories to evaluate
    
    Optional parameters:
    - materiality_threshold: float - Financial materiality threshold
    - automated_testing: bool - Enable automated testing - default: True
    """
    return {
        "success": True,
        "audit_id": audit_id,
        "frameworks_count": len(regulatory_frameworks),
        "business_units_count": len(business_units),
        "lead_auditor": lead_auditor,
        "risk_categories": risk_categories,
        "audit_period": audit_period,
        "automated_testing": automated_testing,
        "status": "Initiated",
        "estimated_completion": "2025-10-13",
        "message": f"Compliance audit {audit_id} initiated successfully"
    }

def book_meeting_room(room_id: str, date: str, start_time: str, end_time: str, attendee_count: int, meeting_title: str, organizer_id: str, equipment_needed: list = None):
    """Reserve a meeting room for a specific date and time.
    
    Required parameters:
    - room_id: str - Unique meeting room ID (e.g., 'CONF-A-301')
    - date: str - Booking date (YYYY-MM-DD)
    - start_time: str - Start time (HH:MM, 24-hour)
    - end_time: str - End time (HH:MM, 24-hour)
    - attendee_count: int - Number of attendees (1-100)
    - meeting_title: str - Title of the meeting
    - organizer_id: str - Employee ID of the organizer
    
    Optional parameters:
    - equipment_needed: list - Equipment required (e.g., ['projector', 'whiteboard'])
    """
    booking_id = f"BOOK_{room_id}_{date}_{start_time.replace(':', '')}"
    
    return {
        "success": True,
        "booking_id": booking_id,
        "room_id": room_id,
        "date": date,
        "start_time": start_time,
        "end_time": end_time,
        "attendee_count": attendee_count,
        "meeting_title": meeting_title,
        "organizer_id": organizer_id,
        "equipment_needed": equipment_needed or [],
        "status": "Confirmed",
        "message": f"Meeting room {room_id} booked successfully for {date}"
    }

def track_shipment(tracking_number: str, carrier: str, include_history: bool = False):
    """Track the current status and location of a shipment.
    
    Required parameters:
    - tracking_number: str - The unique tracking identifier
    - carrier: str - Shipping carrier (FedEx, UPS, DHL, USPS, Amazon Logistics)
    
    Optional parameters:
    - include_history: bool - Whether to include full tracking history - default: False
    """
    tracking_data = {
        "tracking_number": tracking_number,
        "carrier": carrier,
        "status": "In Transit",
        "current_location": "Distribution Center - Chicago, IL",
        "estimated_delivery": "2025-09-15",
        "last_updated": "2025-09-13T08:45:00Z"
    }
    
    if include_history:
        tracking_data["history"] = [
            {"date": "2025-09-12", "location": "Origin Facility", "status": "Picked Up"},
            {"date": "2025-09-13", "location": "Chicago, IL", "status": "In Transit"}
        ]
    
    return {
        "success": True,
        "tracking_info": tracking_data,
        "message": f"Shipment {tracking_number} is currently in transit"
    }

def escalate_support_ticket(ticket_id: str, reason: str, target_team: str, notify_customer: bool = True):
    """Escalate a support ticket to a higher-priority team.
    
    Required parameters:
    - ticket_id: str - Support ticket ID
    - reason: str - Reason for escalation
    - target_team: str - Team to escalate to (senior_support, engineering, billing)
    
    Optional parameters:
    - notify_customer: bool - Notify customer about escalation - default: True
    """
    return {
        "success": True,
        "ticket_id": ticket_id,
        "reason": reason,
        "target_team": target_team,
        "escalated_at": "2025-09-13T10:30:00Z",
        "customer_notified": notify_customer,
        "new_priority": "High",
        "estimated_response": "2025-09-13T14:00:00Z",
        "message": f"Ticket {ticket_id} escalated to {target_team} team"
    }

def create_performance_review(employee_id: str, reviewer_id: str, review_period: dict, overall_rating: int, review_type: str = "annual"):
    """Create a performance review for an employee.
    
    Required parameters:
    - employee_id: str - ID of the employee being reviewed
    - reviewer_id: str - ID of the reviewer (manager/supervisor)
    - review_period: dict - Start and end dates for review period
    - overall_rating: int - Performance rating (1-5 scale)
    
    Optional parameters:
    - review_type: str - Type of review (annual, mid_year, quarterly, probationary) - default: annual
    """
    review_id = f"REV_{employee_id}_{review_type}_{overall_rating}"
    
    return {
        "success": True,
        "review_id": review_id,
        "employee_id": employee_id,
        "reviewer_id": reviewer_id,
        "review_period": review_period,
        "overall_rating": overall_rating,
        "review_type": review_type,
        "status": "Draft",
        "created_date": "2025-09-13",
        "message": f"Performance review {review_id} created with rating {overall_rating}/5"
    }

def schedule_meeting(title: str, start_time: str, end_time: str, participants: list, agenda: str = None, video_call_link: str = None):
    """Schedule a meeting with participants and basic details.
    
    Required parameters:
    - title: str - Meeting title
    - start_time: str - Start time (ISO format)
    - end_time: str - End time (ISO format)
    - participants: list - List of participants with email and role
    
    Optional parameters:
    - agenda: str - Brief agenda
    - video_call_link: str - URL for virtual meeting
    """
    meeting_id = f"MEET_{len(title)}_{len(participants)}"
    
    return {
        "success": True,
        "meeting_id": meeting_id,
        "title": title,
        "start_time": start_time,
        "end_time": end_time,
        "participant_count": len(participants),
        "agenda": agenda or "Agenda to be provided",
        "video_call_link": video_call_link,
        "status": "Scheduled",
        "message": f"Meeting '{title}' scheduled successfully with {len(participants)} participants"
    }

def track_ad_performance(ad_ids: list, platform: str, metrics: list, polling_interval_minutes: int = None):
    """Monitor real-time performance metrics for paid ads across platforms.
    
    Required parameters:
    - ad_ids: list - List of ad IDs to track
    - platform: str - Advertising platform (google_ads, facebook_ads, linkedin_ads)
    - metrics: list - Metrics to fetch (impressions, clicks, cost_per_click, conversion_rate)
    
    Optional parameters:
    - polling_interval_minutes: int - How often to refresh data
    """
    performance_data = {}
    for ad_id in ad_ids:
        performance_data[ad_id] = {
            "impressions": 15420,
            "clicks": 892,
            "cost_per_click": 2.35,
            "conversion_rate": 3.2
        }
    
    return {
        "success": True,
        "platform": platform,
        "ads_tracked": len(ad_ids),
        "metrics_requested": metrics,
        "polling_interval": polling_interval_minutes or 60,
        "performance_data": performance_data,
        "last_updated": "2025-09-13T10:30:00Z",
        "message": f"Tracking {len(ad_ids)} ads on {platform}"
    }

def send_offer_letter(candidate_id: str, job_title: str, salary_base: float, start_date: str, expiration_days: int = 7):
    """Send a job offer letter to a candidate.
    
    Required parameters:
    - candidate_id: str - Candidate's unique ID
    - job_title: str - Job title for the offer
    - salary_base: float - Base salary amount
    - start_date: str - Start date (YYYY-MM-DD)
    
    Optional parameters:
    - expiration_days: int - Days before offer expires - default: 7
    """
    offer_id = f"OFFER_{candidate_id}_{job_title.replace(' ', '_')}"
    
    return {
        "success": True,
        "offer_id": offer_id,
        "candidate_id": candidate_id,
        "job_title": job_title,
        "salary_base": salary_base,
        "start_date": start_date,
        "expiration_date": "2025-09-20",  # start_date + expiration_days
        "status": "Sent",
        "sent_date": "2025-09-13",
        "message": f"Offer letter sent to candidate {candidate_id} for {job_title} position"
    }
```

```python
from openai import OpenAI
import json

from tqdm import tqdm
# from secret_key import openai_api_key
import pandas as pd
import datetime

from openai import OpenAI
import json

client = OpenAI(
  base_url = "https://openrouter.ai/api/v1",
  #api_key = " ",
  api_key = " " #my-own-api-key
  
)


# Utils
def call_model(messages):
    """Call the model with the provided messages"""
    system_message = {
        "role": "system",
        "content": """You are a tool-calling assistant. For every query:

        1. **Select Tool & Check Parameters:**
        * Choose appropriate tool
        * Check for required parameters
        * Use defaults for optional parameters

        2. **Execution Rules:**
        * Execute ONLY when ALL required parameters present
        * NEVER execute with missing required parameters
        * Ask ONLY for missing required parameters
        * Execute immediately once all required parameters received

        Every query must result in a tool call. No direct responses allowed."""
    }
    try:
        response = client.chat.completions.create(
            model="meta-llama/llama-3.1-8b-instruct",
            messages=[system_message] + messages,
            tools=tools,
            temperature=0,
            parallel_tool_calls=False
        )
        return response
    except Exception as e:
        return {"choices": [{"message": {"content": f"Error calling model: {str(e)}"}}]}

# Process single row conversation
def process_single_row_conversation(queries, row_index):
    """Process a single row's conversation and return the conversation history"""
    conversation = []
    messages = []
    
    for query in queries:
        # Add user message
        print(f"Row {row_index + 1} - User: {query['content']}")
        conversation.append({
            "role": "user",
            "content": query['content']
        })
        messages.append({
            "role": "user",
            "content": query['content']
        })
        
        try:
            # Call model with conversation history
            response = call_model(messages)
            
            # Handle Mistral's response structure
            if isinstance(response, dict):
                assistant_message = {
                    "role": "assistant",
                    "content": response.get("choices", [{"message": {"content": "Error processing request"}}])[0]["message"]["content"]
                }
            else:
                # Mistral's response format
                assistant_message = {
                    "role": "assistant",
                    "content": response.choices[0].message.content,
                    "tool_calls": getattr(response.choices[0].message, 'tool_calls', None)
                }
            
            # Process assistant response
            if assistant_message.get('tool_calls'):
                # Format tool calls
                tool_calls_json = []
                for tool_call in assistant_message['tool_calls']:
                    tool_call_json = {
                        "id": tool_call.id,  # Include the tool call ID
                        "type": "function",
                        "function": {
                            "name": tool_call.function.name,
                            "arguments": tool_call.function.arguments
                        }
                    }
                    tool_calls_json.append(tool_call_json)
                
                # Add tool call message to messages (for API)
                messages.append({
                    "role": "assistant",
                    "content": None,
                    "tool_calls": assistant_message['tool_calls']  # Use original tool_calls format
                })
                
                # Add formatted response to conversation (for display)
                tool_call_content = f"<tool_call>{json.dumps(tool_calls_json, indent=2)}</tool_call>"
                print(f"Row {row_index + 1} - Assistant: {tool_call_content}")
                conversation.append({
                    "role": "assistant",
                    "content": tool_call_content
                })
                
                # Add tool response messages for each tool call
                for tool_call in assistant_message['tool_calls']:
                    tool_response = {
                        "role": "tool",
                        "content": "Tool executed successfully",  # Mock response
                        "tool_call_id": tool_call.id
                    }
                    messages.append(tool_response)
                    # Don't add tool responses to conversation display
                
            else:
                # Regular text response
                print(f"Row {row_index + 1} - Assistant: {assistant_message['content']}")
                conversation.append({
                    "role": "assistant",
                    "content": assistant_message['content']
                })
                messages.append({
                    "role": "assistant",
                    "content": assistant_message['content']
                })
            
        except Exception as e:
            print(f"Error processing message: {str(e)}")
            conversation.append({
                "role": "assistant",
                "content": f"Error: {str(e)}"
            })
    
    return conversation

# Main execution
def execute_multi_turn_tool_call(filename):
    """Execute multi-turn tool calls and save each row's conversation separately"""
    try:
        # Load Excel file
        df = pd.read_excel(filename)
        df = df.head(10)  # Process only first 10 rows
        total_rows = len(df)
        print(f"\nFound {total_rows} rows to process (limited to first 10 rows)")

        # Prepare output data
        output_data = []
        
        # Process each row with progress bar
        with tqdm(total=total_rows, desc="Processing rows") as pbar:
            for idx, row in df.iterrows():
                try:
                    # Update progress bar description
                    pbar.set_description(f"Processing row {idx + 1}/{total_rows}")
                    
                    # Parse queries from current row
                    queries = json.loads(row['Queries (3 turns in JSON format)'])

                    # Process conversation for this row
                    conversation = process_single_row_conversation(queries, idx)
                    
                    # Store row data
                    row_data = {
                        'Row_Number': idx + 1,
                        'Original_Queries': row['Queries (3 turns in JSON format)'],
                        'Conversation_History': json.dumps(conversation, indent=2),
                        'Total_Messages': len(conversation)
                    }
                    output_data.append(row_data)
                    
                    print(f"Row {idx + 1} completed - {len(conversation)} messages in conversation\n")
                    
                except Exception as e:
                    print(f"\nError processing row {idx + 1}: {e}")
                    # Add error entry
                    error_data = {
                        'Row_Number': idx + 1,
                        'Original_Queries': str(row.get('Queries (3 turns in JSON format)', 'N/A')),
                        'Conversation_History': json.dumps([{"error": f"Failed to process: {str(e)}"}]),
                        'Total_Messages': 0
                    }
                    output_data.append(error_data)
                
                # Update progress bar
                pbar.update(1)
        
        # Save all conversations to Excel
        output_df = pd.DataFrame(output_data)
        
        # Create Excel file with proper formatting
        with pd.ExcelWriter('llama_miss_para.xlsx', engine='openpyxl') as writer:
            output_df.to_excel(writer, index=False, sheet_name='All_Conversations')
            
            # Format the worksheet
            worksheet = writer.sheets['All_Conversations']
            worksheet.column_dimensions['A'].width = 10  # Row Number
            worksheet.column_dimensions['B'].width = 50  # Original Queries
            worksheet.column_dimensions['C'].width = 100  # Conversation History
            worksheet.column_dimensions['D'].width = 15  # Total Messages
            
            from openpyxl.styles import Alignment
            for row in worksheet.iter_rows():
                for cell in row:
                    cell.alignment = Alignment(wrap_text=True, vertical='top')
        
        print(f"\n✅ Processing completed!")
        print(f"📁 Output saved to: conversation_outputs_by_row.xlsx")
        print(f"📊 Total rows processed: {len(output_data)}")
        print(f"✅ Successful: {len([x for x in output_data if 'error' not in x.get('Conversation_History', '')])}")
        print(f"❌ Errors: {len([x for x in output_data if 'error' in x.get('Conversation_History', '')])}")
        
    except Exception as e:
        print(f"Error loading Excel file: {str(e)}")
        return

if __name__ == "__main__":
    execute_multi_turn_tool_call('D:\\Uni Study\\Sem-6\\6.My Code\\2.Multi_turn\\miss_para_query_dataset.xlsx')
```

```python
from openai import OpenAI
import json
from tqdm import tqdm
import pandas as pd
from datasets import load_dataset

# Initialize OpenAI client with OpenRouter
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=" " #my-own-api-key
)

# Utils
def call_model(messages):
    """Call the model with the provided messages"""
    system_message = {
        "role": "system",
        "content": """You are an AI assistant with access to a set of tools. Your goal is to accurately and helpfully respond to user queries.
        **Workflow:**
        1. **Understand Query:** Analyze the user's request and their intent.
        2. **Decide Tool Use:**
            * If you can answer directly from your knowledge, do so.
            * If external info/action is needed, use a tool.
        3. **Select & Extract Tool Inputs:** 
            * Choose the best tool from `tools` list.
            * Identify all required parameters from the query. 
            * Use default values from the tool specification for any fields not explicitly mentioned by the user.
            * **Always include default values in the tool call, even if the user did not mention them.**
            * Don't make assumptions beyond what's in the query or default values.
        4. **Execute Tool (Output Format):**
            * **If you have all required parameters (from user input or defaults), immediately make the tool call.**
            * **Do not ask for missing optional parameters - use defaults.**
            * **Only ask for clarification if absolutely required parameters are missing.**
            * **Once you receive the missing required parameters, immediately make the tool call without further questions.**
            """
    }
    try:
        response = client.chat.completions.create(
            model="qwen/qwen3-14b",
            messages=[system_message] + messages,
            tools=tools,
            temperature=0,
            parallel_tool_calls=False
        )
        return response
    except Exception as e:
        return {"choices": [{"message": {"content": f"Error calling model: {str(e)}"}}]}

# Process single row conversation
def process_single_row_conversation(queries, row_index):
    """Process a single row's conversation and return the conversation history and tool calls"""
    conversation = []
    tool_calls_gt1 = []
    tool_calls_gt2 = []
    
    for query in queries:
        print(f"Row {row_index + 1} - User: {query['content']}")
        conversation.append({
            "role": "user",
            "content": query['content']
        })
        
        messages = conversation.copy()
        response = call_model(messages)
        
        if isinstance(response, dict):
            assistant_message = {
                "role": "assistant",
                "content": response.get("choices", [{"message": {"content": "Error processing request"}}])[0]["message"]["content"]
            }
        else:
            assistant_message = response.choices[0].message.to_dict()
        
        if assistant_message.get("tool_calls"):
            tool_calls_json = []
            for i, tool_call in enumerate(assistant_message["tool_calls"]):
                tool_call_json = {
                    "name": tool_call["function"]["name"],
                    "arguments": json.loads(tool_call["function"]["arguments"])
                }
                tool_calls_json.append(tool_call_json)
                if i == 0:
                    tool_calls_gt1.append(tool_call_json)
                elif i == 1:
                    tool_calls_gt2.append(tool_call_json)
            
            tool_call_content = f"<tool_call>{json.dumps(tool_calls_json, indent=2)}</tool_call>"
            print(f"Row {row_index + 1} - Assistant: {tool_call_content}")
            conversation.append({
                "role": "assistant",
                "content": tool_call_content
            })
        elif assistant_message["content"]:
            print(f"Row {row_index + 1} - Assistant: {assistant_message['content']}")
            conversation.append({
                "role": "assistant",
                "content": assistant_message['content']
            })
        else:
            print(f"Row {row_index + 1} - Assistant: No response generated")
            conversation.append({
                "role": "assistant",
                "content": "No response generated"
            })
    
    return conversation, tool_calls_gt1, tool_calls_gt2

# Main execution
def execute_multi_turn_tool_call():
    """Execute multi-turn tool calls and save each row's conversation separately"""
    try:
        # Load dataset from Hugging Face
        dataset = load_dataset("kunjanshah/new_multi_turn_function_calling_split", split="test")
        df = pd.DataFrame(dataset)
        total_rows = len(df)
        print(f"\nFound {total_rows} rows to process")
        
        output_data = []
        
        with tqdm(total=total_rows, desc="Processing rows") as pbar:
            for idx, row in df.iterrows():
                try:
                    pbar.set_description(f"Processing row {idx + 1}/{total_rows}")
                    
                    # Parse queries from 'User Queries' column
                    queries = json.loads(row['User Queries'])
                    
                    # Process conversation for this row
                    conversation, tool_calls_gt1, tool_calls_gt2 = process_single_row_conversation(queries, idx)
                    
                    # Store row data
                    row_data = {
                        'id': idx + 1,
                        'Original_Queries': row['User Queries'],
                        'Conversation_History': json.dumps(conversation, indent=2),
                        'Total_Messages': len(conversation),
                        'GT1_Tool_Call': json.dumps(tool_calls_gt1) if tool_calls_gt1 else None,
                        'GT2_Tool_Call': json.dumps(tool_calls_gt2) if tool_calls_gt2 else None
                    }
                    output_data.append(row_data)
                    
                    print(f"Row {idx + 1} completed - {len(conversation)} messages in conversation\n")
                    
                except Exception as e:
                    print(f"\nError processing row {idx + 1}: {e}")
                    error_data = {
                        'id': idx + 1,
                        'Original_Queries': str(row.get('User Queries', 'N/A')),
                        'Conversation_History': json.dumps([{"error": f"Failed to process: {str(e)}"}]),
                        'Total_Messages': 0,
                        'GT1_Tool_Call': None,
                        'GT2_Tool_Call': None
                    }
                    output_data.append(error_data)
                
                pbar.update(1)
        
        # Save to Excel
        output_df = pd.DataFrame(output_data)
        with pd.ExcelWriter('results_qwen-14b.xlsx', engine='openpyxl') as writer:
            output_df.to_excel(writer, index=False, sheet_name='All_Conversations')
            worksheet = writer.sheets['All_Conversations']
            worksheet.column_dimensions['A'].width = 10
            worksheet.column_dimensions['B'].width = 50
            worksheet.column_dimensions['C'].width = 100
            worksheet.column_dimensions['D'].width = 15
            worksheet.column_dimensions['E'].width = 50
            worksheet.column_dimensions['F'].width = 50
            
            from openpyxl.styles import Alignment
            for row in worksheet.iter_rows():
                for cell in row:
                    cell.alignment = Alignment(wrap_text=True, vertical='top')
        
        print(f"\n✅ Processing completed!")
        print(f"📁 Output saved to: openai_multi_turn_results_qwen.xlsx")
        print(f"📊 Total rows processed: {len(output_data)}")
        print(f"✅ Successful: {len([x for x in output_data if 'error' not in x.get('Conversation_History', '')])}")
        print(f"❌ Errors: {len([x for x in output_data if 'error' in x.get('Conversation_History', '')])}")
        
    except Exception as e:
        print(f"Error loading dataset: {str(e)}")
        return

if __name__ == "__main__":
    execute_multi_turn_tool_call()
```
