<details> <summary>📌 Difference between Llama template vs ChatML</summary>

Markers:

ChatML = <|im_start|> / <|im_end|>

Llama = <|start_header_id|> / <|eot_id|>

Token IDs: totally different — Llama tokenizer expects its own IDs.

Training: Llama Instruct models were fine-tuned with the Llama template, not ChatML.

👉 If you feed a Llama model ChatML style (<|im_start|>), the tokenizer will just see normal text tokens, not special “role” tokens. That breaks consistency.

📌 So how to think of it

Alpaca, ShareGPT → early formats, not universal.

ChatML → became a standard in many OpenAI datasets.

Llama template → Meta’s own “ChatML-like” structure, but with different special tokens.

Other variants (Vicuna, Gemma, Zephyr, etc.) → each has their own templates, usually forks of ChatML or Llama style.

✅ In practice:

If your model is Llama-3.2, use the Llama-3.2 template.

Don’t mix ChatML unless you have a conversion step that maps <|im_start|> → <|start_header_id|> etc.

Templates matter because they decide where loss is masked, which tokens mark roles, and how inference behaves.</details>

<details>
<summary>**LoRA Trainable Parameters**</summary>

What controls trainable params in LoRA

r (rank) → linear scale. Bigger r = more params.

target_modules → which linear layers get LoRA.

Model shapes → each layer’s in/out features (from the arch).

#layers → multiplied across all decoder layers.

Not affecting count: lora_alpha, lora_dropout (they change training dynamics, not how many params).

LoRA param formula (per linear layer)

For a linear layer with shape (out_features, in_features), LoRA adds two small matrices:

A: (r, in) and B: (out, r)

Trainable params = r * (in + out)
(plus tiny scales; we ignore them—they’re negligible)

Apply to your Llama-3.2-3B (from your print)

Hidden = 3072, MLP = 8192, heads = 24, KV heads = 8
So per layer:

q_proj: in=3072, out=3072 → 16(3072+3072) = 98,304*

k_proj: in=3072, out=1024 → 16(3072+1024) = 65,536*

v_proj: in=3072, out=1024 → 65,536

o_proj: in=3072, out=3072 → 98,304

gate_proj: in=3072, out=8192 → 16(3072+8192) = 180,224*

up_proj: same → 180,224

down_proj: in=8192, out=3072 → 180,224

Sum per layer:
98,304 + 65,536 + 65,536 + 98,304 + 180,224*3
= 868,352

- The 28 is the number of transformer decoder layers in your model.

```(layers): ModuleList(
  (0-27): 28 x LlamaDecoderLayer(
    ...
  )
)
```

Each of those layers has the same structure:

attention part (q_proj, k_proj, v_proj, o_proj)

feedforward/MLP part (gate_proj, up_proj, down_proj)

- Layers: 28 → 868,352 × 28 = 24,313,856 ✅ (matches your print)
</details>

````python
import pandas as pd
import json
from datasets import Dataset

# Step 3: Formatting function for dataset
def formatting_prompts_func(examples):
    convos = []
    lengths = []

    # Iterate through each conversation in the batch
    for convo_json in examples['Conversation']:
        try:
            # Parse the JSON string from the Conversation column
            convo = json.loads(convo_json)
            convos.append(convo)

            # Tokenize the original conversation to get the length
            inputs = tokenizer.apply_chat_template(convo, tokenize=True, add_generation_prompt=False)
            lengths.append(len(inputs))

        except json.JSONDecodeError:
            # Handle invalid JSON by appending an empty conversation or skipping
            convos.append([])  # Or handle as needed (e.g., log error)
            lengths.append(0) # Append 0 length for invalid conversations


    # Apply chat template to each conversation to get the formatted text
    texts = [tokenizer.apply_chat_template(convo, tokenize=False, add_generation_prompt=False) for convo in convos]

    return {"text": texts, "length": lengths}

# Step 4: Convert DataFrame to Hugging Face Dataset
# dataset = Dataset.from_pandas(df[['Conversation']]) # Removed this line as df is not defined

# Step 5: Apply formatting to dataset
dataset = dataset.map(formatting_prompts_func, batched=True)

# Calculate and print the average length
avg_length = sum(dataset['train']['length']) / len(dataset['train']['length'])
print(f"Average token length: {avg_length}")
print("Char count:", len(dataset['train']['text'][0]))
````

But you did NOT tokenize the text yet. You only formatted it.

🔸What happens in SFTTrainer

When you pass:

it tells the trainer:

“Use this tokenizer to tokenize each "text" entry in the dataset into input IDs and attention masks, and apply smart loss masking (like mask user tokens, train only on assistant tokens).”

🔍 Under the hood:

processing_class must implement the method apply_chat_template(...) and internally connect to PreTrainedTokenizer’s __call__ to tokenize.

The SFTTrainer expects text to be raw strings, and tokenizes it itself using the tokenizer + its template.

🧠 So even if you apply the template manually, you still need to give the tokenizer again so SFTTrainer can tokenize it properly with correct EOS tokens, attention mask, and loss mask.

</details>

That only tells the model how long sequences you intend to use (so rope embeddings or position IDs are sized correctly). It’s like saying:

<details> <summary>📘 <strong>✅ 3. dataset_num_proc = 2 </strong></summary>
This controls how many processes to use for processing the dataset (like .map() calls, tokenization, formatting, etc.).

What it does:

Uses 2 CPU processes in parallel to speed up dataset preprocessing.

Why it matters:

Your dataset has raw conversations → these need to be formatted and tokenized.

If your dataset is large, doing this with num_proc=1 is slow.

With num_proc=2, it runs 2 workers in parallel, so it's ~2× faster.

What other values you can use:

None → defaults to 1 process (slower)

4, 8, 16 → go higher if you have more CPU cores available

🎯 Tip:
Set dataset_num_proc = os.cpu_count() or experiment with 4–8 depending on your machine.

Other values you can set:

dataset_num_proc=4, 8, etc. (depends on your CPU).

Too high? It might crash with memory error.

Safe bet = number of physical CPU cores.

What happens if you don’t set it?

It runs sequentially, which is fine for small data but slow on big datasets.
</details>

<details> <summary>📘 <strong>✅ 4. packing = False </strong></summary>
This means:
"Don’t try to combine multiple short examples into one long token sequence."

Why it matters:

Each of your examples (prompt + response) becomes a token sequence.

Some examples may be very short (e.g., 50 tokens).

If packing=True, it will combine multiple short samples into one long 1024-token chunk.

This wastes fewer tokens, so training is much faster — like 2x to 5x faster.

But:

With packing=True, each training sample might contain multiple conversations → this can confuse the model if you don’t format EOS tokens correctly.

Also can break loss masking (important for SFT).

Why False is safe for now:

You’re doing SFT with loss masking, so each example should stay clean → one prompt → one output → one loss.

packing=False = slower, but more correct for now.
</details>

<details> <summary>📘 <strong>✅ gradient_accumulation_steps = 2 </strong></summary>
What is Gradient Accumulation?
When fine-tuning a neural network (like the Llama model in your notebook), the model learns by adjusting its internal settings (called weights) based on examples it sees. To do this efficiently, it processes multiple examples at once in a "batch." However, large batches require a lot of GPU memory, which can be a problem if your GPU (like the Tesla T4 with 14.741 GB) can’t handle it.
Gradient accumulation is a trick to mimic a larger batch size without needing more GPU memory. Instead of updating the model’s weights after processing one batch, it collects (or "accumulates") the learning updates (called gradients) over multiple smaller batches before applying them.
What Does gradient_accumulation_steps = 2 Mean?

Batch Size: Let’s say your batch size is 8 examples (i.e., batch_size = 8).
Gradient Accumulation Steps = 2: This means the model processes 2 batches (2 × 8 = 16 examples) before updating its weights.
Effective Batch Size: The model behaves as if it processed 16 examples at once (8 × 2 = 16), even though it only used the memory for 8 examples at a time.

Real Meaning in Your Fine-Tuning Setup
In your notebook, you’re fine-tuning a Llama model using a dataset (train-00000-of-00001.parquet) with 1,418 examples for a function-calling task. Here’s what gradient accumulation does:

Saves GPU Memory:

Your Tesla T4 has limited memory (14.741 GB). Processing a large batch (e.g., 16 examples) at once might be too much for the GPU.
With gradient_accumulation_steps = 2, the model processes 8 examples at a time (fitting within memory), collects the learning updates, then processes another 8 examples, and only then updates the weights. This mimics a batch of 16 without crashing the GPU.


Improves Learning:

Larger effective batch sizes (like 16) can make the model’s learning more stable because it sees more examples before deciding how to adjust its weights.
Think of it like studying: reviewing 16 questions before updating your understanding is often better than reviewing just 8, but you still only study 8 at a time to avoid overwhelm.


How It Works in Practice:

For each training step:

Step 1: Process batch 1 (8 examples), calculate gradients, but don’t update weights.
Step 2: Process batch 2 (8 more examples), add gradients to the first batch’s gradients.
Update: Combine the gradients from both steps and update the model’s weights as if it saw 16 examples at once.


This repeats for all 1,418 examples in your dataset, split into batches.



Why Is This Useful for You?

Low GPU Memory: Your setup uses full precision (load_in_4bit=False), which eats up memory. Gradient accumulation lets you achieve a larger effective batch size (16) without needing a bigger GPU.
Flexibility: If you want an even larger effective batch size (e.g., 32), you can increase gradient_accumulation_steps to 4 (8 × 4 = 32). This uses the same memory as a batch of 8 but mimics a batch of 32.
Trade-Off: More accumulation steps mean slower training (because you process more batches before updating), but it’s worth it if memory is tight.

Example in Your Setup

Your Dataset: 1,418 examples, batch_size = 8, gradient_accumulation_steps = 2.
Effective Batch Size: 16 examples.
Training Steps: The dataset is split into 1,418 ÷ 16 ≈ 89 effective batches. Each batch takes 2 GPU steps (2 × 8 examples), so the model processes 2,836 GPU steps total (1,418 × 2).
Impact: The model learns from 16 examples at a time for stable updates, but only uses memory for 8 examples per step, fitting your Tesla T4’s limits.
</details>

<details> <summary>📘 <strong>✅ 1. Where Does 445 Come From?</strong></summary>
445 is the total training steps across all 5 epochs. Each epoch has ~89 steps (1,418 examples ÷ effective batch size 16), so 89 × 5 = 445.

Calculation:

Dataset size: 1,418 examples.
Effective batch size: per_device_train_batch_size × gradient_accumulation_steps = 8 × 2 = 16 examples.
Steps per epoch: Divide the dataset by the effective batch size:
$ 1,418 \div 16 \approx 88.625 $. Since you can’t have a partial step, this is rounded up to 89 steps per epoch.
Total steps across all epochs: With num_train_epochs = 5, the total steps are:
$ 89 \times 5 = 445 $.


What it means: The model processes the entire dataset (1,418 examples) 5 times, and each pass (epoch) takes 89 steps. Each step processes 16 examples (via 2 GPU steps of 8 examples each). The 445 is the total number of steps across all 5 epochs.
In the output: [ 76/445 ... Epoch 0.84/5] means the training is at step 76 out of 445 total steps, and it’s currently at 0.84 epochs (since $ 76 \div 89 \approx 0.84 $).

</details>

<details> <summary>📘 <strong>✅ 2. What is 0.05 in [ 76/445 24:42 < 2:03:13, 0.05 it/s, Epoch 0.84/5]?</strong></summary>
The 0.05 it/s stands for iterations per second, specifically the number of training steps completed per second.

What it means: The model is processing 0.05 steps per second, meaning each step takes about $ 1 \div 0.05 = 20 $ seconds. This is slow, likely due to the Tesla T4’s limited compute power and the model’s size (even with LoRA and fp16 = True).
In the output:

24:42: Time spent so far (24 minutes, 42 seconds) after 76 steps.
< 2:03:13: Estimated time remaining (2 hours, 3 minutes, 13 seconds) to complete the remaining $ 445 - 76 = 369 $ steps.
Why slow?: The Tesla T4 (14.741 GB) is handling a Llama model with max_seq_length = 2048, fp16 = True, and optim = "adamw_8bit". Gradient accumulation (gradient_accumulation_steps = 2) also doubles the time per effective step (two GPU passes per step).


Link to setup: The slow speed is tied to per_device_train_batch_size = 8 and gradient_accumulation_steps = 2, which trade speed for memory efficiency. Using fp16 and adamw_8bit helps fit the model in memory but doesn’t speed up computation much.

</details>

````python
from datasets import load_dataset
from unsloth import FastLanguageModel
import pandas as pd
from tqdm import tqdm
import time
import json
import re

# Load model
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="kunjanshah/fine_tuned_epoch_7_unsloth_qwen3-14b_lora",
    max_seq_length=2048,
    dtype=None,
    load_in_4bit=True,
)
FastLanguageModel.for_inference(model)

# Load test dataset and select first 5 samples
dataset = load_dataset("kunjanshah/new_single_turn_function_calling_split", split="test")
#test_subset = dataset.select(range(min(5, len(dataset))))

results = []

for sample in tqdm(dataset, desc="Generating responses"):
    sample = dict(sample)

    # Parse conversation
    convo = eval(sample["Conversation"]) if isinstance(sample["Conversation"], str) else sample["Conversation"]
    convo = convo[:2]  # Only system and user messages

    # Generate response
    inputs = tokenizer.apply_chat_template(
        convo,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt",
    ).to("cuda")

    start_time = time.time()
    outputs = model.generate(input_ids=inputs, max_new_tokens=2048, use_cache=True)
    exec_time = time.time() - start_time

    response = tokenizer.batch_decode(outputs, skip_special_tokens=False)[0]

    # Extract assistant response
    start_tag = "<|im_start|>assistant"
    end_tag = "<|im_end|>"

    if start_tag in response:
        assistant_start = response.find(start_tag) + len(start_tag)
        response = response[assistant_start:]

        if end_tag in response:
            response = response.split(end_tag)[0]

    response = response.strip()

    # Extract tool call if present
    tool_call = None
    tool_call_start = "<tool_call>"
    tool_call_end = "</tool_call>"

    if tool_call_start in response and tool_call_end in response:
        tool_start_idx = response.find(tool_call_start) + len(tool_call_start)
        tool_end_idx = response.find(tool_call_end)
        tool_call_text = response[tool_start_idx:tool_end_idx].strip()

        try:
            tool_call = json.loads(tool_call_text)
            # Remove tool call from response text
            response = response.replace(f"{tool_call_start}{tool_call_text}{tool_call_end}", "").strip()
        except json.JSONDecodeError:
            tool_call = None

    # Clean up response text - remove special tags and extra whitespace
    response = re.sub(r'<think>.*?</think>', '', response, flags=re.DOTALL)
    response = re.sub(r'<[^>]+>', '', response)  # Remove any other XML-like tags
    response = re.sub(r'\s+', ' ', response).strip()  # Normalize whitespace

    # Token usage
    input_tokens = inputs.shape[1]
    output_tokens = outputs.shape[1] - input_tokens

    results.append({
        "model_response": response,
        "tool_call": tool_call,
        "actual_call": sample.get("model_response", sample.get("function_call")),
        "input_tokens": input_tokens,
        "output_tokens": output_tokens,
        "execution_time": exec_time,
    })

# Save results
df = pd.DataFrame(results)
df.to_csv("fine_tuned_epoch_7_unsloth_qwen3-14b_lora_.csv", index=False)
print(df.head())
````
